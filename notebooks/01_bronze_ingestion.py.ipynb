## 01 - Bronze Ingestion

Ingestão bruta dos 8 CSVs do dataset Olist como tabelas `Delta` na camada Bronze.

Sem transformação, apenas leitura do `Volume` e escrita como Delta Table,

com inferência de `schema`.

```sql
%python

tables = [
    "customers",
    "orders",
    "order_items",
    "order_payments",
    "order_reviews",
    "products",
    "sellers",
    "geolocation",
]

base_path = "/Volumes/olist_project/bronze/raw_files" # Troque por seu caminho

for t in tables:
    df = spark.read.csv(
        f"{base_path}/olist_{t}_dataset.csv",
        header=True,
        inferSchema=True,
    )
    df.write.format("delta").mode("overwrite").saveAsTable(
        f"olist_project.bronze.{t}"
    )
    print(f"{t}: {df.count()} linhas ingeridas")
```
