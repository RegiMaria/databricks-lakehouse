## 01 - Bronze Ingestion

Ingestão bruta dos 8 CSVs do dataset Olist como tabelas `Delta` na camada Bronze.

Sem transformação, apenas leitura do `Volume` e escrita como Delta Table,

com inferência de `schema`.

### 1. Ingestão de arquivos e escrita em tabela Delta

```sql
%python
# Configuração (variáveis) para ingestão de dados
tables = [
    "customers",
    "orders",
    "order_items",
    "order_payments",
    "order_reviews",
    "products",
    "sellers",
    "geolocation",
]

base_path = "/Volumes/olist_project/bronze/raw_files"

for t in tables:
    # Leitura do CSV
    df = spark.read.csv(
        f"{base_path}/olist_{t}_dataset.csv",
        header=True,
        inferSchema=True,
    )

    # Escrita como Delta Table
    (
        df.write.format("delta")
        .mode("overwrite")
        .saveAsTable(f"olist_project.bronze.{t}")
    )

    # Confirmação
    print(f"{t}: {df.count()} linhas ingeridas")
```

### 2. Validação via SQL

```sql
SHOW TABLES IN olist_project.bronze;
```
